## Section A: RAG Fundamentals (Q1-Q10)

---

**Q1.** What does RAG stand for?

a) Rapid Automated Generation  
b) Retrieval-Augmented Generation  
c) Real-time AI Generation  
d) Retrieval and Augmentation Gateway  

<details><summary>✅ Answer</summary>

**b) Retrieval-Augmented Generation**  
RAG enhances LLM responses by first retrieving relevant documents from a knowledge base, then using those documents as context for generation.
</details>

---

**Q2.** Which of these problems does RAG NOT solve?

a) LLM hallucination  
b) Knowledge cutoff  
c) Changing the model's writing style permanently  
d) Lack of access to private data  

<details><summary>✅ Answer</summary>

**c) Changing the model's writing style permanently**  
That's what **fine-tuning** does. RAG provides external knowledge at inference time but doesn't modify the model itself.
</details>

---

**Q3.** A healthcare company wants to build a chatbot that answers questions about their proprietary drug database that is updated daily. What approach should they use?

a) Fine-tune an LLM on the drug database  
b) Use RAG with the drug database as the knowledge source  
c) Use a larger LLM that might already know the information  
d) Use prompt engineering alone  

<details><summary>✅ Answer</summary>

**b) Use RAG**  
The data is: (1) proprietary/private, (2) updated daily. RAG handles both perfectly — it retrieves current data without retraining. Fine-tuning would require daily retraining, which is impractical.
</details>

---

**Q4.** What are the two main phases of a RAG pipeline?

a) Training and Inference  
b) Encoding and Decoding  
c) Indexing and Querying  
d) Input and Output  

<details><summary>✅ Answer</summary>

**c) Indexing and Querying**  
Indexing (offline): Load → Chunk → Embed → Store.  
Querying (online): Embed Query → Retrieve → Augment Prompt → Generate.
</details>

---

**Q5.** Which is the correct order of stages in the RAG indexing phase?

a) Embed → Load → Chunk → Store  
b) Load → Embed → Chunk → Store  
c) Load → Chunk → Embed → Store  
d) Chunk → Load → Store → Embed  

<details><summary>✅ Answer</summary>

**c) Load → Chunk → Embed → Store**  
You must load documents first, then split them into chunks, then convert chunks to embeddings, and finally store in a vector database.
</details>

---

**Q6.** When would you choose fine-tuning OVER RAG?

a) When you need answers from private documents  
b) When you need to change the model's output format and style  
c) When your data changes frequently  
d) When you need source attribution  

<details><summary>✅ Answer</summary>

**b) When you need to change the model's output format and style**  
Fine-tuning modifies the model's weights to change its behavior, style, or format. RAG only adds external knowledge, not behavioral changes.
</details>

---

**Q7.** What does "grounding" mean in the context of RAG?

a) Training the model on ground truth data  
b) Ensuring responses are based on retrieved evidence  
c) Converting text to vector representations  
d) Storing data in a database  

<details><summary>✅ Answer</summary>

**b) Ensuring responses are based on retrieved evidence**  
Grounding means the LLM's answer is anchored in actual retrieved documents, not its own training data or imagination.
</details>

---

**Q8.** What does "augmented" refer to in RAG?

a) The model is made larger  
b) The training data is augmented with more examples  
c) The prompt is enhanced with retrieved context  
d) The output is augmented with formatting  

<details><summary>✅ Answer</summary>

**c) The prompt is enhanced with retrieved context**  
The user's query is "augmented" by adding relevant documents as context before sending to the LLM.
</details>

---

**Q9.** Which framework is most commonly used for building RAG pipelines?

a) TensorFlow  
b) PyTorch  
c) LangChain  
d) Scikit-learn  

<details><summary>✅ Answer</summary>

**c) LangChain**  
LangChain provides abstractions for all RAG components: loaders, splitters, embeddings, vector stores, retrievers, and chains.
</details>

---

**Q10.** A RAG system retrieves correct documents but generates wrong answers. Which component is most likely the problem?

a) Document loader  
b) Vector database  
c) Prompt template or LLM  
d) Text splitter  

<details><summary>✅ Answer</summary>

**c) Prompt template or LLM**  
If retrieval is correct, the issue is in the generation stage — either the prompt doesn't properly instruct the LLM to use the context, or the LLM itself is too weak.
</details>

## Section B: Data Parsing & Chunking (Q11-Q20)

---

**Q11.** Which LangChain loader would you use for a PDF file?

a) TextLoader  
b) CSVLoader  
c) PyPDFLoader  
d) JSONLoader  

<details><summary>✅ Answer</summary>

**c) PyPDFLoader**  
PyPDFLoader is designed specifically for PDF files and creates one Document per page.
</details>

---

**Q12.** A Document object in LangChain has which two attributes?

a) text and labels  
b) content and tags  
c) page_content and metadata  
d) data and schema  

<details><summary>✅ Answer</summary>

**c) page_content and metadata**  
`page_content` holds the text, `metadata` is a dictionary with source info, page numbers, etc.
</details>

---

**Q13.** You need to extract tables from a complex PDF. Which parser is best?

a) PyPDFLoader  
b) TextLoader  
c) pdfplumber  
d) CSVLoader  

<details><summary>✅ Answer</summary>

**c) pdfplumber**  
pdfplumber specializes in table extraction from PDFs. For very complex layouts with images, LlamaParse or Docling would also be good choices.
</details>

---

**Q14.** What is the default text splitter recommended for most RAG pipelines?

a) CharacterTextSplitter  
b) TokenTextSplitter  
c) RecursiveCharacterTextSplitter  
d) SemanticChunker  

<details><summary>✅ Answer</summary>

**c) RecursiveCharacterTextSplitter**  
It tries multiple separators in order (paragraphs → lines → sentences → words) to keep chunks semantically coherent.
</details>

---

**Q15.** What happens when chunk_size is set too small (e.g., 50 characters)?

a) Better retrieval precision  
b) Fragments lose meaning and context  
c) Vector database runs out of space  
d) LLM generates longer responses  

<details><summary>✅ Answer</summary>

**b) Fragments lose meaning and context**  
Chunks become too small to carry meaningful information, leading to poor embeddings and noisy retrieval.
</details>

---

**Q16.** What is the purpose of chunk_overlap?

a) To increase storage efficiency  
b) To prevent context loss at chunk boundaries  
c) To speed up embedding generation  
d) To reduce the number of chunks  

<details><summary>✅ Answer</summary>

**b) To prevent context loss at chunk boundaries**  
Overlap ensures sentences spanning chunk boundaries are preserved in both chunks. Recommended: 10-20% of chunk_size.
</details>

---

**Q17.** What is the recommended chunk_size and chunk_overlap for general-purpose RAG?

a) chunk_size=50, chunk_overlap=10  
b) chunk_size=1000, chunk_overlap=200  
c) chunk_size=5000, chunk_overlap=0  
d) chunk_size=100, chunk_overlap=100  

<details><summary>✅ Answer</summary>

**b) chunk_size=1000, chunk_overlap=200**  
This provides a good balance between context and precision. Overlap is 20% of chunk_size.
</details>

---

**Q18.** What is the difference between `split_text()` and `split_documents()`?

a) split_text is faster  
b) split_documents preserves metadata  
c) split_text works with PDFs  
d) They are identical  

<details><summary>✅ Answer</summary>

**b) split_documents preserves metadata**  
`split_text(string)` → list of strings. `split_documents(docs)` → list of Documents with metadata preserved.
</details>

---

**Q19.** How does semantic chunking differ from RecursiveCharacterTextSplitter?

a) Semantic chunking is faster  
b) Semantic chunking uses fixed-size chunks  
c) Semantic chunking splits based on meaning changes using embeddings  
d) They are the same  

<details><summary>✅ Answer</summary>

**c) Semantic chunking splits based on meaning changes using embeddings**  
It produces variable-size chunks based on topic shifts detected by embedding similarity. It's slower but higher quality.
</details>

---

**Q20.** The default separators in RecursiveCharacterTextSplitter are:

a) `[".", ",", " "]`  
b) `["\n\n", "\n", ". ", " ", ""]`  
c) `[" ", "\n"]`  
d) `["\n\n", "\n"]`  

<details><summary>✅ Answer</summary>

**b) `["\n\n", "\n", ". ", " ", ""]`**  
It tries paragraphs first, then lines, then sentences, then words, and finally characters as a last resort.
</details>

## Section C: Embeddings & Vector Databases (Q21-Q30)

---

**Q21.** What do embeddings represent?

a) Compressed file formats  
b) Numerical vectors capturing semantic meaning  
c) Database indexes  
d) Encrypted text  

<details><summary>✅ Answer</summary>

**b) Numerical vectors capturing semantic meaning**  
Similar texts produce similar vectors that are close in vector space, enabling semantic search.
</details>

---

**Q22.** What dimension does `all-MiniLM-L6-v2` produce?

a) 128  
b) 384  
c) 768  
d) 1536  

<details><summary>✅ Answer</summary>

**b) 384**  
all-MiniLM-L6-v2 is a compact, free model from HuggingFace producing 384-dimensional vectors.
</details>

---

**Q23.** You indexed documents with a 384d model. You try to query with a 768d model. What happens?

a) It works but slower  
b) Dimension mismatch error  
c) Results are less accurate  
d) The system auto-converts  

<details><summary>✅ Answer</summary>

**b) Dimension mismatch error**  
The indexing and querying models MUST produce vectors of the same dimension. You'd need to re-index with the new model.
</details>

---

**Q24.** Which distance metric does FAISS use by default?

a) Cosine similarity  
b) Dot product  
c) L2 (Euclidean) distance  
d) Manhattan distance  

<details><summary>✅ Answer</summary>

**c) L2 (Euclidean) distance**  
In FAISS, lower scores = more similar. This is opposite to cosine similarity where higher = more similar.
</details>

---

**Q25.** Which vector database auto-persists data without manual save?

a) FAISS  
b) ChromaDB  
c) Neither  
d) Both  

<details><summary>✅ Answer</summary>

**b) ChromaDB**  
ChromaDB auto-persists when given a `persist_directory`. FAISS requires explicit `save_local()` calls.
</details>

---

**Q26.** For a production app with billions of vectors, which vector DB is best?

a) FAISS  
b) ChromaDB  
c) Pinecone  
d) SQLite  

<details><summary>✅ Answer</summary>

**c) Pinecone**  
Pinecone is a fully managed cloud service designed for production scale with billions of vectors, auto-scaling, and managed infrastructure.
</details>

---

**Q27.** Which vector database has the best metadata filtering capabilities?

a) FAISS  
b) ChromaDB  
c) Qdrant  
d) They're all equal  

<details><summary>✅ Answer</summary>

**c) Qdrant**  
Qdrant supports complex filter expressions with rich payload, including range queries, nested filters, and boolean combinations.
</details>

---

**Q28.** What does `allow_dangerous_deserialization=True` do in FAISS?

a) Enables faster search  
b) Allows loading pickled FAISS indexes (potential security risk)  
c) Encrypts the vector data  
d) Enables GPU support  

<details><summary>✅ Answer</summary>

**b) Allows loading pickled FAISS indexes**  
FAISS uses Python's pickle for serialization, which can execute arbitrary code. This flag is a safety acknowledgment.
</details>

---

**Q29.** What is the difference between `embed_query()` and `embed_documents()`?

a) embed_query is for batch processing  
b) embed_query embeds one string, embed_documents embeds a list  
c) They produce different dimension vectors  
d) embed_documents is deprecated  

<details><summary>✅ Answer</summary>

**b) embed_query embeds one string, embed_documents embeds a list**  
Some models also optimize differently for queries vs documents (asymmetric embedding).
</details>

---

**Q30.** Cosine similarity of 0.95 between two vectors means:

a) They are very different  
b) They are moderately similar  
c) They are very semantically similar  
d) There's an error  

<details><summary>✅ Answer</summary>

**c) They are very semantically similar**  
Cosine similarity ranges from -1 to 1. 0.95 indicates the vectors point in nearly the same direction.
</details>

## Section D: Retrievers (Q31-Q40)

---

**Q31.** What is the default search_type for `vectorstore.as_retriever()`?

a) mmr  
b) similarity  
c) bm25  
d) hybrid  

<details><summary>✅ Answer</summary>

**b) similarity**  
Returns the top-k documents most similar to the query vector.
</details>

---

**Q32.** Your RAG system returns 3 documents that all say essentially the same thing. Which retriever type would fix this?

a) Basic similarity  
b) BM25  
c) MMR  
d) Compression  

<details><summary>✅ Answer</summary>

**c) MMR (Maximal Marginal Relevance)**  
MMR balances relevance with diversity, ensuring retrieved documents cover different aspects of the topic.
</details>

---

**Q33.** In MMR, what is the relationship between `k` and `fetch_k`?

a) k should equal fetch_k  
b) fetch_k should be larger than k  
c) k should be larger than fetch_k  
d) They are unrelated  

<details><summary>✅ Answer</summary>

**b) fetch_k should be larger than k**  
fetch_k is the candidate pool (e.g., 20), from which k diverse documents (e.g., 5) are selected.
</details>

---

**Q34.** A user searches for "FAISS library" but the dense retriever returns documents about "similarity search" instead. What would fix this?

a) Increase chunk_size  
b) Use hybrid search (dense + BM25)  
c) Change the LLM  
d) Decrease k  

<details><summary>✅ Answer</summary>

**b) Use hybrid search (dense + BM25)**  
Dense retrieval focuses on semantic meaning and may miss exact keywords. BM25 (sparse) excels at keyword matching. Combining both catches both.
</details>

---

**Q35.** What does the Ensemble Retriever use to merge results from multiple retrievers?

a) Simple averaging  
b) Reciprocal Rank Fusion (RRF)  
c) Max pooling  
d) Weighted sum of scores  

<details><summary>✅ Answer</summary>

**b) Reciprocal Rank Fusion (RRF)**  
RRF scores documents based on their rank positions across multiple retrievers. Documents ranking high in multiple retrievers get the highest combined scores.
</details>

---

**Q36.** What does BM25 retrieval excel at compared to dense retrieval?

a) Understanding synonyms  
b) Semantic understanding  
c) Exact keyword matching  
d) Multi-language search  

<details><summary>✅ Answer</summary>

**c) Exact keyword matching**  
BM25 is based on term frequency and directly rewards documents containing the exact search terms. It misses synonyms but catches exact matches.
</details>

---

**Q37.** When would you use a Multi-Query Retriever?

a) When you need faster retrieval  
b) When the query is ambiguous or complex  
c) When you have too many documents  
d) When you need metadata filtering  

<details><summary>✅ Answer</summary>

**b) When the query is ambiguous or complex**  
It uses an LLM to generate multiple query variations, retrieves for each, and merges results — maximizing recall for complex queries.
</details>

---

**Q38.** What does metadata filtering do in retrieval?

a) Removes metadata from documents  
b) Filters documents before similarity search based on metadata values  
c) Adds metadata to query  
d) Sorts results by metadata  

<details><summary>✅ Answer</summary>

**b) Filters documents before similarity search based on metadata values**  
For example, only search in documents where `department="engineering"` or `date > "2024-01-01"`.
</details>

---

**Q39.** The Contextual Compression Retriever is useful when:

a) Vectors are too large  
b) Retrieved chunks contain mix of relevant and irrelevant content  
c) You need faster searches  
d) You have too few documents  

<details><summary>✅ Answer</summary>

**b) Retrieved chunks contain mix of relevant and irrelevant content**  
It uses an LLM to extract only the relevant portions, reducing noise in the context sent to the final LLM.
</details>

---

**Q40.** Which statement about dense vs sparse retrieval is correct?

a) Dense retrieval uses TF-IDF  
b) Sparse retrieval captures semantic meaning  
c) Dense captures meaning, sparse captures keywords  
d) They produce the same results  

<details><summary>✅ Answer</summary>

**c) Dense captures meaning, sparse captures keywords**  
Dense (embedding-based) captures semantic similarity ("happy" ≈ "joyful"). Sparse (BM25/TF-IDF) captures exact keyword frequency.
</details>

## Section E: Chains, Prompting & Evaluation (Q41-Q52)

---

**Q41.** What is the default chain_type in RetrievalQA?

a) map_reduce  
b) refine  
c) stuff  
d) map_rerank  

<details><summary>✅ Answer</summary>

**c) stuff**  
The stuff chain type concatenates all retrieved documents into a single prompt. It's the simplest and works when context fits in the window.
</details>

---

**Q42.** You have 50 large documents that exceed the LLM's context window. Which chain_type should you use?

a) stuff  
b) map_reduce  
c) Simple prompt  
d) None — increase context window  

<details><summary>✅ Answer</summary>

**b) map_reduce**  
map_reduce processes each document independently with the LLM, then combines summaries. This avoids context window overflow.
</details>

---

**Q43.** What does the `|` (pipe) operator do in LCEL?

a) Performs logical OR  
b) Chains components: output of one becomes input of next  
c) Runs components in parallel  
d) Filters results  

<details><summary>✅ Answer</summary>

**b) Chains components: output of one becomes input of next**  
`prompt | llm | parser` means: format prompt → send to LLM → parse output.
</details>

---

**Q44.** What does `RunnablePassthrough()` do in an LCEL chain?

a) Skips the component  
b) Passes input through unchanged  
c) Caches the result  
d) Validates the input  

<details><summary>✅ Answer</summary>

**b) Passes input through unchanged**  
Used to forward the user's question directly while other parts of the chain process it differently (e.g., retriever gets context).
</details>

---

**Q45.** What is the difference between zero-shot and few-shot prompting?

a) Zero-shot is faster  
b) Few-shot provides examples to guide output format  
c) Zero-shot uses more tokens  
d) They are identical  

<details><summary>✅ Answer</summary>

**b) Few-shot provides examples to guide output format**  
Zero-shot: instruction only. Few-shot: instruction + examples of desired input-output pairs.
</details>

---

**Q46.** What temperature should you use for a factual RAG QA system?

a) 1.0  
b) 0.7  
c) 0.0  
d) 0.5  

<details><summary>✅ Answer</summary>

**c) 0.0**  
Temperature 0 makes the model deterministic and factual — essential for RAG where accuracy matters over creativity.
</details>

---

**Q47.** Which RAGAS metric measures if the answer is grounded in the context?

a) Answer Relevance  
b) Context Precision  
c) Faithfulness  
d) Context Recall  

<details><summary>✅ Answer</summary>

**c) Faithfulness**  
Faithfulness measures whether the claims in the answer are supported by the retrieved context. Low faithfulness = hallucination.
</details>

---

**Q48.** Your RAG system has high Context Recall but low Context Precision. What does this mean?

a) Not finding relevant documents  
b) Finding all relevant docs but also many irrelevant ones  
c) LLM is hallucinating  
d) Embeddings are broken  

<details><summary>✅ Answer</summary>

**b) Finding all relevant docs but also many irrelevant ones**  
High recall = finding everything relevant. Low precision = also retrieving lots of noise. Fix: reduce k, use MMR, or add metadata filtering.
</details>

---

**Q49.** How do you get source documents back from RetrievalQA?

a) Set `verbose=True`  
b) Set `return_source_documents=True`  
c) Call `get_sources()`  
d) Sources are always returned  

<details><summary>✅ Answer</summary>

**b) Set `return_source_documents=True`**  
Then access via `result['source_documents']` which returns a list of Document objects.
</details>

---

**Q50.** Which output parser would you use to get structured JSON from an LLM?

a) StrOutputParser  
b) XMLOutputParser  
c) JsonOutputParser  
d) DictOutputParser  

<details><summary>✅ Answer</summary>

**c) JsonOutputParser**  
It parses the LLM's string output into a Python dictionary. Use StrOutputParser for plain text.
</details>

---

**Q51.** A user reports that after changing the embedding model, the vector database returns no results. What's the likely issue?

a) The LLM is too slow  
b) Embedding dimension mismatch — need to re-index  
c) The prompt is wrong  
d) The vector DB is full  

<details><summary>✅ Answer</summary>

**b) Embedding dimension mismatch — need to re-index**  
The old vectors have different dimensions than the new query vectors. All documents must be re-indexed with the new embedding model.
</details>

---

**Q52.** Which combination gives the BEST retrieval quality for a production RAG system?

a) Similarity search only  
b) BM25 only  
c) Ensemble (dense + BM25) with reranking  
d) Random sampling  

<details><summary>✅ Answer</summary>

**c) Ensemble (dense + BM25) with reranking**  
Dense catches semantic similarity, BM25 catches keywords, and reranking re-scores using a cross-encoder for maximum precision. This is the gold standard for production RAG.
</details>
